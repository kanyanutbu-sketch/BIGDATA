```python
import sys, re, collections
import pandas as pd
import matplotlib.pyplot as plt

REPO = "alpindale/two-million-bluesky-posts"
N_FILES = 3  # ใช้ 1-3 ไฟล์ตามโจทย์

```

```python
# ----------------- Your code here -----------------
# ใบ้: อ่าน JSON lines ทีละบรรทัดด้วย generator ไม่ต้องโหลดทั้งไฟล์
# def iter_jsonl(path):
#     with open(path, encoding='utf-8') as f:
#         for line in f:
#             yield json.loads(line)

def read_any(p):
    print("อ่านไฟล์:", p)
    if p.endswith(".parquet"):
        return pd.read_parquet(p)
    try:
        return pd.read_json(p, lines=True)
    except ValueError:
        return pd.read_json(p)


def load_data():
    import os
    # กรองอาร์กิวเมนต์ของ Jupyter/Colab (เช่น -f kernel.json) ออก เหลือเฉพาะไฟล์ข้อมูลจริง
    args = [a for a in sys.argv[1:]
            if os.path.isfile(a) and a.endswith((".parquet", ".json", ".jsonl"))
            and "kernel" not in os.path.basename(a)]
    if args:
        return pd.concat([read_any(a) for a in args], ignore_index=True)

    from huggingface_hub import list_repo_files, hf_hub_download
    files = sorted(f for f in list_repo_files(REPO, repo_type="dataset")
                   if f.startswith("posts_") and f.endswith(".jsonl"))
    print(f"พบไฟล์ข้อมูล {len(files)} ไฟล์")
    # เลือก N_FILES ไฟล์กระจายตามช่วงเวลา (ต้น-กลาง-ท้าย) เพื่อให้ครอบคลุมหลายชั่วโมง
    idx = sorted({round(i * (len(files) - 1) / max(N_FILES - 1, 1)) for i in range(N_FILES)})
    chosen = [files[i] for i in idx]
    print("ใช้ไฟล์:", chosen)
    paths = [hf_hub_download(REPO, f, repo_type="dataset") for f in chosen]
    return pd.concat([read_any(x) for x in paths], ignore_index=True)


df = load_data()
print("\nคอลัมน์:", list(df.columns))
df["text"] = df["text"].fillna("")
# format="ISO8601" รองรับ timestamp หลายรูปแบบ; errors=coerce กันค่าเพี้ยน

# ---------- ตรวจจับภาษา (dataset จริงไม่มีคอลัมน์ predicted_language) ----------
if "predicted_language" not in df.columns:
    import py3langid as langid
    CLEAN = re.compile(r"https?://\S+|\S+\.\w{2,4}/\S*|@\S+|#\S+")
    def detect_lang(t):
        c = CLEAN.sub(" ", t).strip()
        if len(re.findall(r"[\u3040-\u30ff]", c)) >= 1: return "ja"
        if len(re.findall(r"[\uAC00-\uD7AF]", c)) >= 2: return "ko"
        if len(re.findall(r"[\u0E00-\u0E7F]", c)) >= 2: return "th"
        if len(re.findall(r"[\u4e00-\u9fff]", c)) >= 2: return "zh"
        if len(re.sub(r"[\W\d_]+", "", c)) < 12: return "und"   # สั้น/อีโมจิ/ลิงก์ล้วน
        return langid.classify(c[:300])[0]
    print("กำลังตรวจจับภาษา (อาจใช้เวลา 1-3 นาที)...")
    df["predicted_language"] = df["text"].map(detect_lang)

df["ts"] = pd.to_datetime(df["created_at"], utc=True, errors="coerce", format="ISO8601")
n_bad = df["ts"].isna().sum()
```

```python
print("\n===== Q1: คนโพสต์เมื่อไร =====")
print(f"จำนวนโพสต์ทั้งหมด: {len(df):,}  (parse เวลาไม่ได้ {n_bad:,})")
print(f"ช่วงเวลา (ทั้งหมด): {df.ts.min()}  ->  {df.ts.max()}")
# ข้อมูลมีโพสต์ที่ claim เวลาย้อนหลัง (เช่น 2011) จึงดูช่วงหลักด้วย (1%-99%)
q_lo, q_hi = df.ts.quantile(0.01), df.ts.quantile(0.99)
print(f"ช่วงที่ครอบคลุม 98% ของโพสต์: {q_lo} -> {q_hi}")
print("โพสต์ที่ระบุปีก่อน 2024 (ส่วนใหญ่น่าจะ import/ย้อนหลัง): "
      f"{(df.ts.dt.year < 2024).sum():,} ({(df.ts.dt.year < 2024).mean():.2%})")

main = df[(df.ts >= q_lo) & (df.ts <= q_hi)].copy()
main["hour_utc"] = main.ts.dt.hour
main["hour_ict"] = main.ts.dt.tz_convert("Asia/Bangkok").dt.hour
h_utc = main.hour_utc.value_counts().sort_index().reindex(range(24), fill_value=0)
h_ict = main.hour_ict.value_counts().sort_index().reindex(range(24), fill_value=0)
print(f"ชั่วโมงที่โพสต์มากสุด (UTC): {h_utc.idxmax():02d}:00  ({h_utc.max():,} โพสต์)")
print(f"ชั่วโมงที่โพสต์มากสุด (เวลาไทย): {h_ict.idxmax():02d}:00  ({h_ict.max():,} โพสต์)")
print("Top 3 ชั่วโมง (UTC):", h_utc.nlargest(3).to_dict())
if main.ts.dt.date.nunique() <= 2:
    print("หมายเหตุ: ข้อมูลครอบคลุมช่วงสั้นมาก จึงบอกแพทเทิร์น 'ชั่วโมงในวัน' ไม่ได้เต็มที่")
```

```python
# ---------- Q2: ภาษา ----------
print("\n===== Q2: ภาษา =====")
if "predicted_language" in df.columns:
    lang = df["predicted_language"].fillna("unknown")
else:
    print("ไม่มีคอลัมน์ predicted_language -> ใช้วิธีเดาจากสคริปต์ตัวอักษรแบบหยาบ ๆ")
    def guess(t):
        if re.search(r"[\u0E00-\u0E7F]", t): return "tha_Thai"
        if re.search(r"[\u3040-\u30ff]", t): return "jpn_Jpan"
        if re.search(r"[\uAC00-\uD7AF]", t): return "kor_Hang"
        if re.search(r"[\u4e00-\u9fff]", t): return "zho_Hani"
        if re.search(r"[\u0400-\u04FF]", t): return "cyrillic"
        if re.search(r"[\u0600-\u06FF]", t): return "arabic_script"
        return "latin(other)"
    lang = df.text.map(guess)
und_share = (lang == "und").mean()
print(f"โพสต์ที่ระบุภาษาไม่ได้ (สั้น/อีโมจิ/ลิงก์ล้วน/ข้อความว่าง) = {und_share:.1%} -> ตัดออกก่อนคิดสัดส่วน")
lang = lang[lang != "und"]
lang_share = lang.value_counts(normalize=True)
lang_cnt = lang.value_counts()
top_lang = pd.DataFrame({"posts": lang_cnt.head(10), "share_%": (lang_share.head(10) * 100).round(2)})
print(top_lang)
print(f"Top-3 รวมกัน = {lang_share.head(3).sum():.1%}")
if "language_confidence" in df.columns:
    print(f"ค่าเฉลี่ย confidence ของ model = {df.language_confidence.mean():.2f}; "
          f"โพสต์ที่ confidence < 0.5 = {(df.language_confidence < 0.5).mean():.1%}")
```

```python
# ---------- Q3: พูดถึงอะไร ----------
print("\n===== Q3: Hashtag และคำเด่น =====")
HT = re.compile(r"#(\w+)", re.UNICODE)
tags = collections.Counter()
for t in df.text:
    for h in set(x.lower() for x in HT.findall(t)):
        tags[h] += 1
top_tags = tags.most_common(15)
print("Top 15 hashtags (นับต่อโพสต์):")
for h, c in top_tags:
    print(f"  #{h}: {c:,}")

STOP = set("""the a an and or but to of in on for with is it this that i you he she we they at as be are was were
my your our me so not no just have has had do did do's if from by about up out all can will would get got like
im its it's i'm don't rt https http www com amp one more what when how who there their then than them too very
que die und der das ist nicht ich mit den von zu les des est pas une dans pour qui por una los las con del para como pero mas más pra uma não nao com se um em os das dos ein eine auch sich auf sie
""".split())
eng = df[df["predicted_language"].isin(["en", "eng_Latn"])] if "predicted_language" in df.columns else df
words = collections.Counter()
for t in eng.text:
    t = re.sub(r"https?://\S+|\S+\.\w{2,4}/\S*|@\S+|#\S+", " ", t.lower())
    ws = [w for w in re.findall(r"[a-z']{3,}", t) if w not in STOP]
    words.update(ws)
    words.update(f"{a} {b}" for a, b in zip(ws, ws[1:]))  # bigram
print("\nTop 20 คำ/วลี (เฉพาะโพสต์ภาษาอังกฤษ ตัด stopword):")
for w, c in words.most_common(20):
    print(f"  {w}: {c:,}")

print("\nตัวอย่างข้อความจริงของ hashtag เด่น (ตัดความยาว 200 ตัวอักษร ไม่แสดงตัวตนผู้เขียน):")
for h, _ in top_tags[:3]:
    ex = df[df.text.str.lower().str.contains("#" + re.escape(h) + r"(?!\w)", regex=True)]["text"].head(2)
    for e in ex:
        print(f"  [#{h}] {e[:200]!r}")
```

```python
# ---------- Q4: Visualization 1 รูป ----------
fig, ax = plt.subplots(1, 3, figsize=(18, 5))
ax[0].bar(h_ict.index, h_ict.values, color="#2b7de9")
ax[0].set_title("Posts by hour (Thailand time, UTC+7)")
ax[0].set_xlabel("Hour of day"); ax[0].set_ylabel("Posts")
ax[0].set_xticks(range(0, 24, 3))
top_l = lang_share.head(8) * 100
ax[1].barh(top_l.index[::-1], top_l.values[::-1], color="#e0793b")
ax[1].set_title("Top languages (% of posts)"); ax[1].set_xlabel("%")
th = top_tags[:10]
ax[2].barh([f"#{h}" for h, _ in th][::-1], [c for _, c in th][::-1], color="#3aa17e")
ax[2].set_title("Top 10 hashtags (posts)"); ax[2].set_xlabel("Posts")
fig.suptitle(f"Bluesky sample: {len(df):,} posts", fontsize=14)
plt.tight_layout()
plt.savefig("bluesky_overview.png", dpi=150)  # เก็บสำรองไว้แนบรายงาน
plt.show()  # แสดงรูปในหน้า Colab/Jupyter

# ข้อมูลประกอบข้อจำกัด
print("\n===== ตัวเลขช่วยเขียนข้อจำกัด =====")
print(f"สัดส่วนโพสต์ที่เป็น reply: {df.reply_to.notna().mean():.1%}")
print(f"สัดส่วนโพสต์ที่มีรูป: {df.has_images.mean():.1%}")
print(f"โพสต์ข้อความว่าง: {(df.text.str.strip() == '').mean():.1%}")
print(f"จำนวนผู้เขียนไม่ซ้ำ: {df.author.nunique():,}; "
      f"top 1% ของผู้เขียนโพสต์รวม {df.author.value_counts().head(max(1, df.author.nunique() // 100)).sum() / len(df):.1%}")
```
